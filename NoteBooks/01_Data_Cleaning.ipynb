```python
# ============================================================
# PHASE 1 : DATA UNDERSTANDING & DATA CLEANING
# Rossmann Store Sales Forecasting
# ============================================================

import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)


# Load Datasets

train = pd.read_csv("/content/train.csv", low_memory=False)
store = pd.read_csv("/content/store.csv", low_memory=False)


# Dataset Overview

print("="*60)
print("TRAIN DATASET")
print("="*60)
print(train.head())

print("\nShape:", train.shape)

print("\nInfo:")
print(train.info())

print("\nMissing Values:")
print(train.isnull().sum())

print("\nDuplicate Rows:", train.duplicated().sum())

print("="*60)
print("STORE DATASET")
print("="*60)
print(store.head())

print("\nShape:", store.shape)

print("\nInfo:")
print(store.info())

print("\nMissing Values:")
print(store.isnull().sum())

print("\nDuplicate Rows:", store.duplicated().sum())


# Missing Value Percentage

print("\nTrain Missing Percentage")
print((train.isnull().sum()/len(train))*100)

print("\nStore Missing Percentage")
print((store.isnull().sum()/len(store))*100)


# Convert Date

train["Date"] = pd.to_datetime(train["Date"])

# Merge Datasets

df = train.merge(store, on="Store", how="left")

print("\nMerged Shape:", df.shape)


# Missing Values After Merge

print("\nMissing Values After Merge")
print(df.isnull().sum().sort_values(ascending=False))


# Handle Missing Values


# Numerical
df["CompetitionDistance"] = df["CompetitionDistance"].fillna(
    df["CompetitionDistance"].median()
)

# Competition
df["CompetitionOpenSinceMonth"] = df["CompetitionOpenSinceMonth"].fillna(0)
df["CompetitionOpenSinceYear"] = df["CompetitionOpenSinceYear"].fillna(0)

# Promo2
df["Promo2SinceWeek"] = df["Promo2SinceWeek"].fillna(0)
df["Promo2SinceYear"] = df["Promo2SinceYear"].fillna(0)

# Promo Interval
df["PromoInterval"] = df["PromoInterval"].fillna("None")

# Feature Engineering


df["Year"] = df["Date"].dt.year

df["Month"] = df["Date"].dt.month

df["Week"] = df["Date"].dt.isocalendar().week.astype(int)

df["Day"] = df["Date"].dt.day

df["Quarter"] = df["Date"].dt.quarter

df["DayName"] = df["Date"].dt.day_name()

df["Weekend"] = df["DayName"].isin(["Saturday","Sunday"]).astype(int)


# Final Dataset Check

print("="*60)
print("FINAL DATASET")
print("="*60)

print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nSummary Statistics")
print(df.describe())

print("\nFirst Five Rows")
print(df.head())


# Save Clean Dataset

df.to_csv("cleaned_sales.csv", index=False)

print("\n✅ Clean dataset saved as cleaned_sales.csv")
```
